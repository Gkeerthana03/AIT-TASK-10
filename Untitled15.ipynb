```python
from kanren import Relation, facts, var, run, conde


parent = Relation()


facts(
    parent,
    ("Abe", "Homer"),
    ("Homer", "Bart"),
    ("Homer", "Lisa"),
)


def grandparent(x, z):
    y = var()
    return conde((parent(x, y), parent(y, z)))


x = var()
result = run(1, x, grandparent(x, "Bart"))

print(f"The grandparent of Bart is: {result[0]}")
```

```python
from kanren import Relation, facts, var, run


is_animal = Relation()
has_color = Relation()


facts(is_animal, "Frog", "Crocodile", "Canary", "Flamingo")

facts(
    has_color,
    ("Frog", "Green"),
    ("Crocodile", "Green"),
    ("Canary", "Yellow"),
    ("Flamingo", "Pink")
)


x = var()


green_animals = run(0, x, is_animal(x), has_color(x, "Green"))
print(f"Green animals: {green_animals}")


color = var()
canary_color = run(1, color, has_color("Canary", color))
print(f"A canary is: {canary_color[0]}")
```

```python
from kanren import Relation, facts, var, run


studies = Relation()
likes = Relation()


facts(
    studies,
    ("Arun", "Python"),
    ("Bala", "Java"),
    ("Cathy", "Python"),
    ("David", "C++")
)

facts(
    likes,
    ("Arun", "Cricket"),
    ("Bala", "Football"),
    ("Cathy", "Chess"),
    ("David", "Cricket")
)


x = var()


python_students = run(0, x, studies(x, "Python"))
print(f"Students studying Python: {python_students}")


subject = var()
bala_subject = run(1, subject, studies("Bala", subject))
print(f"Bala studies: {bala_subject[0]}")


cricket_students = run(0, x, likes(x, "Cricket"))
print(f"Students who like Cricket: {cricket_students}")
```
